```python
%sql
create table if not exists table_work.basketball.player_stats as 
select season, games_played, minutes_played, TRB, AST, PTS 
from table_work.default.sheet_1_table_1
```

```python
%sql
select * from table_work.basketball.player_stats
```

```python
%sql
select season, pts 
from table_work.basketball.player_stats
order by pts desc
limit 1 
```

```python
%sql
select season, minutes_played, pts, round(pts/minutes_played,2) as points_per_minute
from table_work.basketball.player_stats
order by points_per_minute desc
limit 10
```

```python
%sql
select sum(games_played) * sum(minutes_played) as total_minutes_played, sum(pts) * sum(games_played) as total_points, round(total_points/total_minutes_played, 2) as average_points_per_minute
from table_work.basketball.player_stats
```
